```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import jn_zeros, jnp_zeros
plt.style.use('bmh')
```

## Ceros de las funciones de Bessel de orden entero $J_\nu(\alpha_{\nu,n})=0$:

En el módulo `scipy.special` las raíces de las funciones de Bessel están implementadas en la función `jn_zeros`. En particular `jn_zeros(nu,N)` entrega un arreglo con las primeras $N$ raíces de la función $J_\nu$.

```python
for nu in range(5):
    print('nu = '+str(nu)+': '+str(jn_zeros(nu,5)))
```

Podemos graficar estas raíces, de la forma siguiente:

```python
nmax = 20
numax = 10
for nu in range(numax):
    plt.scatter(nu*np.ones(nmax),jn_zeros(nu,nmax), s=12, color='blue')
plt.ylim(0,50)
plt.xlim(-1,numax)
plt.xlabel('$\\nu$', fontsize=15)
plt.ylabel('$\\alpha_{\\nu,n}$', fontsize=15)
plt.xticks(range(numax))
plt.grid(True)
plt.savefig('../figs/fig-Bessel-ceros-01.pdf')
```

Note que estas raíces no están igualmente espaciadas, lo que se aprecia mejor al graficar la _diferencia entre una raiz y la siguiente_:

```python
plt.figure()
nmax = 20
numax = 10
for nu in range(numax):
    plt.plot(range(nmax-1),np.diff(jn_zeros(nu,nmax)), marker='o', markersize=3, lw=1, label='$\\nu = %d$'%nu)
plt.xlim(-1,nmax-1)
plt.ylim(3,)
plt.xlabel('$n$', fontsize=15)
plt.ylabel('$\\alpha_{\\nu,n+1}-\\alpha_{\\nu,n}$', fontsize=15)
plt.xticks(range(nmax))
plt.legend()
plt.grid(True)
plt.savefig('../figs/fig-Bessel-ceros-02.pdf')
```

Vemos sin embargo que la diferencia $\alpha_{\nu,n+1}-\alpha_{\nu,n}$ tiende a $\pi$ para valores grandes de $n$.

## Ceros de la derivada de las funciones de Bessel de orden entero $J'_\nu(\alpha_{\nu,n})=0$:

Análogamente, las raíces de las derivadas $J'_\nu$ son calculadas por la función `jnp_zeros` de `scipy.special`:

```python
for nu in range(5):
    print('nu = '+str(nu)+': '+str(jnp_zeros(nu,5)))
```

```python
plt.figure()
nmax = 20
numax = 10
for nu in range(numax):
    plt.scatter(nu*np.ones(nmax),jnp_zeros(nu,nmax), s=12, color='blue')
plt.ylim(0,50)
plt.xlim(-1,numax)
plt.xlabel('$\\nu$', fontsize=15)
plt.ylabel('$\\beta_{\\nu,n}$', fontsize=15)
plt.xticks(range(numax))
plt.grid(True)
```

En este caso, el gráfico de las diferencias es también similar al caso anterior:

```python
plt.figure()
nmax = 20
numax = 10
for nu in range(numax):
    plt.plot(range(nmax-1),np.diff(jnp_zeros(nu,nmax)), marker='o', markersize=3, lw=1, label='$\\nu = %d$'%nu)
plt.xlim(-1,nmax-1)
plt.ylim(3,)
plt.xlabel('$n$', fontsize=15)
plt.ylabel('$\\beta_{\\nu,n+1}-\\beta_{\\nu,n}$', fontsize=15)
plt.xticks(range(nmax))
plt.legend()
plt.grid(True)
```
